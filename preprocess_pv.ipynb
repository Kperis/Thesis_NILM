```python
import pandas as pd

df = pd.read_csv('datasets/pv_data/pv_data.csv', parse_dates=['Valid Date UTC'])
df.shape
```

```python
df.head()
```

```python
from dateutil.relativedelta import relativedelta

df['Valid Date UTC'] = df['Valid Date UTC'].apply(lambda x: x - relativedelta(months=18))
```

```python
print(df['Valid Date UTC'].min(), df['Valid Date UTC'].max())
```

```python
df = df.rename(columns={
    'Valid Date UTC': 'timestamps',
    'Energy KWh': 'total_energy'
})

df = df[['timestamps', 'total_energy']]

```

```python
df['total_energy'] = df['total_energy'].apply(lambda x: x if x >= 75 else 0.0)
df['is_active'] = df['total_energy'].apply(lambda x: 1 if x > 0 else 0)

df.to_csv('datasets/pv_data/shifted_pv_data.csv', index=False)
```
